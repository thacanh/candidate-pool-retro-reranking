This notebook reads compact public summaries only. It does not fit models or rerun inference/statistical resampling. Figure 1 and private manuscript assets are excluded. Change cosmetic settings below, not scientific selectors. Outputs go to `figures_rebuilt/`. Install the package with analysis extras and Jupyter first: `python -m pip install -e '.[analysis]' jupyter`.


```python
from pathlib import Path
import sys, hashlib
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
assert (ROOT / 'outputs/transfer_analysis/figure_inputs/adaptation.csv').is_file(), 'Open this notebook from the public repository.'
sys.path.insert(0, str(ROOT / 'src'))
from rerank.figures import plot_figures as figures
for line in (ROOT / 'checksums.sha256').read_text().splitlines():
    expected, name = line.split('  ', 1)
    if name.startswith('outputs/'):
        assert hashlib.sha256((ROOT / name).read_bytes()).hexdigest() == expected, name
print('Frozen numerical inputs verified; cosmetic notebook edits are allowed.')
OUT = ROOT / 'figures_rebuilt'
```
